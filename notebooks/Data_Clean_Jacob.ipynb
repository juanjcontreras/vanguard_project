```python
import yaml
import pandas as pd

with open("../config.yaml", "r", encoding="utf-8") as file:
    config = yaml.safe_load(file)

df_demo    = pd.read_csv(config["input_data"]["demo"])
df_web_pt1 = pd.read_csv(config["input_data"]["web_pt1"])
df_web_pt2 = pd.read_csv(config["input_data"]["web_pt2"])
df_roster  = pd.read_csv(config["input_data"]["roster"])
```

```python
#create a clean copy
df_clean = df_demo.copy()
```

```python
df_clean.shape
```

```python
#check missing values
df_clean.isna().sum()
```

```python
#To see the rows containing missing values
df_clean[df_clean.isna().any(axis=1)]
```

```python
#drop all missing values
important_columns = [
    "clnt_tenure_yr",
    "clnt_tenure_mnth",
    "clnt_age",
    "gendr",
    "num_accts",
    "bal",
    "calls_6_mnth",
    "logons_6_mnth"
]

df_clean = df_clean.dropna(
    subset=important_columns,
    how="all"
)
```

```python
df_clean[df_clean["clnt_age"].isna()]
```

```python
median_age = df_clean["clnt_age"].median()

df_clean["clnt_age"] = df_clean["clnt_age"].fillna(median_age)
```

```python
df_clean.isna().sum()
```

```python
df_clean.head(9583)
```

```python
df_clean["gendr"].value_counts(dropna=False)
```

```python
df_clean["gendr"] = (
    df_clean["gendr"]
    .str.strip()
    .str.upper()
    .replace("X", "U")
)
```

```python
df_clean["gendr"].value_counts()
```

```python
df_clean.duplicated().sum()
```

```python
integer_columns = [
    "clnt_tenure_yr",
    "clnt_tenure_mnth",
    "num_accts",
    "calls_6_mnth",
    "logons_6_mnth"
]

df_clean[integer_columns] = df_clean[integer_columns].astype(int)
```
